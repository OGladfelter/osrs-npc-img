# Delete flagged images

Reads `../data/flagged_junk_images.csv` (produced by `flag_junk_images.ipynb`, after visual
review) and deletes the corresponding chathead/body files from `../data/chatheads/` and
`../data/bodies/`.

Share `flagged_junk_images.csv` with groupmates — everyone runs this notebook against their
own copy of `data/` to prune the same set of images.

```python
import os
import pandas as pd

flagged = pd.read_csv("../data/flagged_junk_images.csv")
print(f"{len(flagged)} images flagged for deletion")
flagged["type"].value_counts()
```

```python
flagged.head(10)
```

```python
FOLDER = {"chathead": "chatheads", "body": "bodies"}

deleted, missing = 0, 0
for _, row in flagged.iterrows():
    path = f"../data/{FOLDER[row['type']]}/{row['id']}.png"
    if os.path.exists(path):
        os.remove(path)
        deleted += 1
    else:
        missing += 1

print(f"Deleted {deleted} files, {missing} were already missing")
```
